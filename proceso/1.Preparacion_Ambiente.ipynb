```python
%sql
create widget text storageName default ".";
create widget text catalogwk default "catalog_prod";
```

```python
%python
storageName = dbutils.widgets.get("storageName")
```

```python
%sql
CREATE CATALOG IF NOT EXISTS $catalogwk
COMMENT 'Catalogo para la arquitectura medallion del ambiente de dev';
```

```python
%sql
DROP SCHEMA IF EXISTS $catalogwk.raw;
DROP SCHEMA IF EXISTS $catalogwk.bronze;
DROP SCHEMA IF EXISTS $catalogwk.silver;
DROP SCHEMA IF EXISTS $catalogwk.golden;
```

```python
%python
dbutils.fs.rm(f"s3://s3smartdata2109/prod/bronze",True)
dbutils.fs.rm(f"s3://s3smartdata2109/prod/silver",True)
dbutils.fs.rm(f"s3://s3smartdata2109/prod/golden",True)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS $catalogwk.raw;
CREATE SCHEMA IF NOT EXISTS $catalogwk.bronze;
CREATE SCHEMA IF NOT EXISTS $catalogwk.silver;
CREATE SCHEMA IF NOT EXISTS $catalogwk.golden;
```

###Tablas Bronze

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogwk.bronze.circuits (
circuit_id integer,
circuit_ref string,
name string,
location string,
country string,
latitude double,
longitude double,
altitude integer,
ingestion_date timestamp
)
USING DELTA
LOCATION "s3://${storageName}/prod/bronze/circuits"
```

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogwk.bronze.races (
race_id integer,
race_year integer,
round integer,
circuit_id integer,
name string,
ingestion_date timestamp
)
USING DELTA
PARTITIONED BY (race_year)
LOCATION "s3://${storageName}/prod/bronze/races"
```

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogwk.bronze.constructors (
  constructor_id integer,
  constructor_ref string,
  name string,
  nationality string,
  ingestion_date timestamp
)
USING DELTA
LOCATION "s3://${storageName}/prod/bronze/constructors"
```

###Tablas Silver

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogwk.silver.circuits_transformed (
  race_id integer,
  race_year integer,
  round integer,
  circuit_id integer,
  name_race string,
  ingestion_date timestamp,
  circuit_ref string,
  name string,
  location string,
  country string,
  latitude double,
  longitude double,
  altitude integer,
  altitude_category string,
  years_diferences integer,
  lat_diff integer,
  race_type string,
  near_equator string
)
USING DELTA
LOCATION "s3://${storageName}/prod/silver/circuits_transformed"
```

###Tablas Golden

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogwk.golden.golden_raced_partitioned (
  race_year integer,
  conteo long,
  max_altitude integer,
  min_altitude integer,
  country string,
  race_type string,
  near_equator string
)
USING DELTA
LOCATION "s3://${storageName}/prod/golden/golden_raced_partitioned"
```
